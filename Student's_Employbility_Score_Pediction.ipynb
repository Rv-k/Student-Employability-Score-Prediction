### One-Way ANOVA and Eta Squared Analysis

This analysis evaluates whether categorical features (`Gender`, `Major`, `GitHub_Profile`,`Academic_Performance`,`English_Proficiency` and `LinkedIn_Profile`) have a statistically significant relationship with the continuous variable `Employability_Score`.

* **F-statistic** measures the variation between groups relative to variation within groups.
* **P-value** determines statistical significance. A p-value < 0.05 indicates a significant difference between group means.
* **Eta Squared (η²)** measures the proportion of variance in `Employability_Score` explained by the categorical feature.
* A higher η² indicates a stronger effect of the categorical feature on `Employability_Score`.

This analysis helps identify categorical features that may be useful for predicting `Employability_Score`.


```python
from scipy.stats import f_oneway

categorical_features = [
    "Gender",
    "Major",
    "GitHub_Profile",
    "LinkedIn_Profile",
    "Academic_Performance",
    "English_Proficiency"
]

for col in categorical_features:

    groups = [
        group["Employability_Score"].values
        for _, group in df.groupby(col)
    ]

    f_stat, p_value = f_oneway(*groups)

    # Grand mean
    grand_mean = df["Employability_Score"].mean()

    # Between-group sum of squares
    ss_between = sum(
        len(group) * (np.mean(group) - grand_mean) ** 2
        for group in groups
    )

    # Total sum of squares
    ss_total = np.sum(
        (df["Employability_Score"] - grand_mean) ** 2
    )

    eta_squared = ss_between / ss_total

    print(f"{col}")
    print(f"F-statistic : {f_stat:.4f}")
    print(f"P-value    : {p_value:.6f}")
    print(f"Eta squared (η²): {eta_squared:.6f}")
    print("-" * 30)
```

# Variance Inflation Factor (VIF) Analysis

VIF is used to detect multicollinearity among the selected numerical features. A higher VIF indicates stronger correlation between a feature and the other features.

* VIF < 5 → Generally acceptable
* VIF ≥ 5 → Potential multicollinearity
* VIF ≥ 10 → High multicollinearity

This analysis helps determine whether the selected numerical features can be retained for model building without significant multicollinearity.

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
import pandas as pd

selected_num_features = [
    "CGPA",
    "Programming_Skill",
    "Internships",
    "Resume_Score",
    "Communication_Skills",
    "Problem_Solving",
    "Interview_Score"
]

X_num = df[selected_num_features]

vif = pd.DataFrame()
vif["Feature"] = X_num.columns
vif["VIF"] = [
    variance_inflation_factor(X_num.values, i)
    for i in range(X_num.shape[1])
]

print(vif.sort_values("VIF", ascending=False))
```

# 6. Skewness¶
What skewness is: a measure of how lopsided a numeric column's distribution is. A value near 0 means roughly symmetric (bell-shaped); a large positive or negative value means the data leans heavily to one side, with a long tail.

Why it matters: models like Linear Regression assume features are roughly well-behaved. A heavily skewed column (think: a long tail of extreme values) can quietly drag the model's predictions in that direction. Tree-based models like Random Forest don't care about skew — but since we're also training a Linear Regression baseline, it's worth fixing.

* |Skewness| < 0.5       → approximately symmetric
* 0.5 ≤ |Skewness| < 1   → moderately skewed
* |Skewness| ≥ 1         → highly skewed

# 8. Train-Test Split

```python
from sklearn.model_selection import train_test_split

normal_col = ['Major'] # perform One Hot Encoding
ordinal_col = ['GitHub_Profile', 'Academic_Performance', 'English_Proficiency'] # perform Ordinal Encoding
numeric_col = ["CGPA", "Programming_Skill", "Internships", "Resume_Score", "Communication_Skills", "Problem_Solving", "Interview_Score"] # perform Stander Scaler

feature_col = normal_col + ordinal_col + numeric_col
X = df[feature_col]
y = df['Employability_Score']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

# 9. Preprocessing using ColumnTransformer

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OrdinalEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer

normal_col_pipeline = Pipeline(steps=[
    ('normal encode', OneHotEncoder(handle_unknown='ignore'))
])


ordinal_col_pipeline = Pipeline(steps=[
    ('one hot encode', OrdinalEncoder(
     categories=[
        ["No", "Yes"],                         # GitHub_Profile
        ["Poor", "Average", "Good", "Excellent"],  # Academic_Performance
        ["Basic", "Intermediate", "Advanced"]     # English_Proficiency
     ]
    ))
])

numeric_col_pipeline = Pipeline(steps=[
    ('standard scaler', StandardScaler())
])


preprocessor = ColumnTransformer(transformers=[
    ('Normal', normal_col_pipeline, normal_col),
    ('Ordinal', ordinal_col_pipeline, ordinal_col),
    ('Numeric', numeric_col_pipeline, numeric_col)
])
```

# 10. Model Building
# 10.1 — Baseline: Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
lr_model_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressior', LinearRegression())
])

lr_model_pipeline.fit(X_train, y_train)
lr_model_predict_test = lr_model_pipeline.predict(X_test)
lr_model_predict_train = lr_model_pipeline.predict(X_train)

lr_model_test_r2_score = r2_score(y_test,lr_model_predict_test)
lr_model_train_r2_score = r2_score(y_train,lr_model_predict_train)
lr_model_mae = mean_absolute_error(y_test,lr_model_predict_test)
lr_model_mse = np.sqrt(mean_squared_error(y_test,lr_model_predict_test))

print(f"R² Score of Testing {lr_model_test_r2_score}")
print(f"R² Score of Training {lr_model_train_r2_score}")
print(f"Mean absolute error {lr_model_mae}")
print(f"Root mean squared error {lr_model_mse}")
```

# 10.2 - Random Forest

```python
from sklearn.ensemble import RandomForestRegressor
rf_model_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('random forest', RandomForestRegressor(random_state=42))
])

rf_model_pipeline.fit(X_train, y_train)
rf_model_predict_test = rf_model_pipeline.predict(X_test)
rf_model_predict_train = rf_model_pipeline.predict(X_train)

rf_model_test_r2_score = r2_score(y_test,rf_model_predict_test)
rf_model_train_r2_score = r2_score(y_train,rf_model_predict_train)
rf_model_mae = mean_absolute_error(y_test,rf_model_predict_test)
rf_model_mse = np.sqrt(mean_squared_error(y_test,rf_model_predict_test))

print(f"R² Score of Testing {rf_model_test_r2_score}")
print(f"R² Score of Training {rf_model_train_r2_score}")
print(f"Mean absolute error {rf_model_mae}")
print(f"Root mean squared error {rf_model_mse}")
```

# 11. Model Evaluation

```python
model_evaluation = pd.DataFrame({
    'Model'            : ['Linear Regression', 'Random Forest'],
    'R2 Score Testing' :  [lr_model_test_r2_score, rf_model_test_r2_score],
    'R2 Score Training': [lr_model_train_r2_score, rf_model_train_r2_score],
    'MAE'              : [lr_model_mae, rf_model_mae],
    'RMSE'              : [lr_model_mse, rf_model_mse]
})

model_evaluation
```
